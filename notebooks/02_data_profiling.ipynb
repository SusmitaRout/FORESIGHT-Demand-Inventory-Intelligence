```python
import pandas as pd
from pathlib import Path
```

```python
# Project paths
DATA_PATH = Path("../data/raw")
```

```python
# Load datasets
sku_master = pd.read_csv(DATA_PATH / "sku_master.csv")
calendar = pd.read_csv(DATA_PATH / "calendar.csv")
sales_daily = pd.read_csv(DATA_PATH / "sales_daily.csv")
inventory_snapshots = pd.read_csv(DATA_PATH / "inventory_snapshots.csv")

print("All four datasets loaded successfully!")
```

```python
datasets = {
    "SKU Master": sku_master,
    "Calendar": calendar,
    "Sales Daily": sales_daily,
    "Inventory Snapshots": inventory_snapshots
}

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)
    print(df.dtypes)
```

```python
for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)
    print(df.isnull().sum())
```

```python
for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

```python
sales_daily["date"] = pd.to_datetime(sales_daily["date"])
inventory_snapshots["date"] = pd.to_datetime(inventory_snapshots["date"])

print("Sales period:", sales_daily["date"].min(), "to", sales_daily["date"].max())
print("Inventory period:", inventory_snapshots["date"].min(), "to", inventory_snapshots["date"].max())
```

```python
# Check the grain of each dataset

print("SKU Master:")
print("Rows:", len(sku_master))
print("Unique SKUs:", sku_master["sku"].nunique())

print("\nCalendar:")
print("Rows:", len(calendar))
print("Unique dates:", calendar["date"].nunique())

print("\nSales Daily:")
print("Rows:", len(sales_daily))
print("Unique SKUs:", sales_daily["sku"].nunique())
print("Unique dates:", sales_daily["date"].nunique())

print("\nInventory Snapshots:")
print("Rows:", len(inventory_snapshots))
print("Unique SKUs:", inventory_snapshots["sku"].nunique())
print("Unique dates:", inventory_snapshots["date"].nunique())
```

```python
# Data Quality Summary

quality_summary = []

for name, df in datasets.items():
    quality_summary.append({
        "Dataset": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Missing Values": int(df.isnull().sum().sum()),
        "Duplicate Rows": int(df.duplicated().sum())
    })

quality_report = pd.DataFrame(quality_summary)

quality_report
```
